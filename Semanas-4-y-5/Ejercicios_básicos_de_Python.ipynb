```python
# Pedir nombre del estudiante
nombre_estudiante = input("Por favor, ingrese el nombre del estudiante: ")

# Pedir 3 calificaciones
suma = 0
promedio = 0
for i in range(3):
  calificacion = float(input(f"Ingrese la {i+1}º calificación: "))
  suma = suma + calificacion

# Calcular el promedio
promedio = suma / 3

# Mostrar el promedio
print(f"\nEl promedio de {nombre_estudiante} es: {promedio:.2f}.")

# Mostrar si aprobó o reprobó
if promedio >= 70:
  print("Aprobó.")
else:
  print("Reprobó.")

# Mostrar mensaje si el promedio es mayor a 90
if promedio > 90:
  print("¡Felicidades, el promedio es mayor a 90!")

```
