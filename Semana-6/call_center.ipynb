```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("call_center.csv", sep=";", encoding="cp1252")
df
```

```python
df['Tiempo_Espera_Min'] = df['Tiempo_Espera_Min'].astype(str).str.replace(',', '.').astype(float)
a=df['Tiempo_Espera_Min'].mean()
b=df['Tiempo_Espera_Min'].median()
c=df['Tiempo_Espera_Min'].mode()

print(f'Media del tiempo es: {a:.2f}',f'Mediana del tiempo es: {b}',f'y la o las Modas del tiempo son: {c}')
```

```python
varianza=df['Tiempo_Espera_Min'].var()
desviacion=df['Tiempo_Espera_Min'].std()
minimo = df['Tiempo_Espera_Min'].min()
maximo = df['Tiempo_Espera_Min'].max()
print(f"La varianza es: {varianza:.2f}")
print(f"La desviación estándar es: {desviacion:.2f}")
print(f"El valor mínimo es: {minimo:.2f}")
print(f"El valor máximo es: {maximo:.2f}")

```

```python

q1 = np.percentile(df['Tiempo_Espera_Min'], 25)
q2 = np.percentile(df['Tiempo_Espera_Min'], 50)
q3 = np.percentile(df['Tiempo_Espera_Min'], 75)

print(q1, q2, q3)
```

```python
plt.figure(figsize=(12, 7))

sns.histplot(df['Tiempo_Espera_Min'], kde=True, bins=10, color='skyblue', edgecolor='black')

plt.title('Histograma de Tiempo de Espera', fontsize=16)
plt.xlabel('Tiempo de Espera (min)', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.grid(axis='y', alpha=0.75)

# Marcar cuartiles
plt.axvline(q1, color='green', linestyle='--', linewidth=2, label=f'Q1: {q1:.2f}')
plt.axvline(q2, color='red', linestyle='-', linewidth=2, label=f'Q2 (Mediana): {q2:.2f}')
plt.axvline(q3, color='purple', linestyle='--', linewidth=2, label=f'Q3: {q3:.2f}')

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
plt.tight_layout()
plt.show()
```
