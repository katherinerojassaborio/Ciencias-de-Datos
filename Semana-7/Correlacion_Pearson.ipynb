```python
import pandas as pd
from scipy import stats

df = pd.DataFrame({
    'comercio': ['A', 'B', 'C', 'D', 'E', 'F', 'G'],
    'x': [2, 4, 1, 6, 3, 5, 2],
    'y': [45, 68, 30, 95, 52, 80, 40]
})

r, p = stats.pearsonr(df.x, df.y)
print(f"\nCorrelación Pearson r = {r:.4f},  p = {p:.4f}")
```

```python
import pandas as pd
from scipy import stats

comercios_input = input("Introduce los nombres de los comercios (separados por comas): ")
x_input = input("Introduce los valores de x (separados por comas): ")
y_input = input("Introduce los valores de y (separados por comas): ")


comercios = [nombre.strip() for nombre in comercios_input.split(',')]
x = [float(valor.strip()) for valor in x_input.split(',')]
y = [float(valor.strip()) for valor in y_input.split(',')]

df = pd.DataFrame({
    'comercio': comercios,
    'x': x,
    'y': y
})

# Correlación de Pearson
r, _ = stats.pearsonr(df.x, df.y)
print(f"Correlación Pearson r = {r:.4f}")

# La guaria, La amistad, Minisuper Eli, Abismo, Megasuper, Bendicion, La Cabaña
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.figure(figsize=(12, 7))

# A. Seaborn regplot: Puntos y línea de tendencia automática
chart = sns.regplot(data=df, x='x', y='y',
                    scatter_kws={'s': 20, 'color': 'darkblue'},
                    line_kws={'color': 'red', 'linewidth': 2},
                    ci =None)


for i in range(df.shape[0]):
    plt.text(x=df.x[i], y=df.y[i] + 2,
             s=df.comercio[i],
             fontdict=dict(color='black', size=10),
             horizontalalignment='center'),


plt.title('Análisis de Correlación de Pearson entre Visitas en el Mes y Monto por Comercio', fontsize=16)
plt.xlabel('Visitas en el mes', fontsize=12)
plt.ylabel('Monto de pedido', fontsize=12)


stats_text = f'Correlación r = {r:.4f}'
plt.gca().text(0.05, 0.95, stats_text, transform=plt.gca().transAxes, fontsize=12,
               verticalalignment='top', bbox=dict(boxstyle='round,pad=0.5', facecolor='white', alpha=0.8, edgecolor='gray'))


plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

datos = {
    "Comercio": ["A", "B", "C", "D", "E", "F", "G"],
    "Visitas": [2, 4, 1, 6, 3, 5, 2],
    "Pedido": [45, 68, 30, 95, 52, 80, 40]
}

df = pd.DataFrame(datos)

print(df)

# Calcular correlación de Pearson
correlacion = df["Visitas"].corr(df["Pedido"])

print("\nCoeficiente de correlación:", correlacion)

plt.figure(figsize=(8, 8))

plt.pie(
    df["Pedido"],
    labels=df["Comercio"],
    autopct="%1.1f%%",
    startangle=90,
    explode=[0.03] * len(df),
    shadow=True
)

plt.title(
    "Participación del monto de pedidos por comercio",
    fontsize=15,
    fontweight="bold"
)

plt.show()
```
