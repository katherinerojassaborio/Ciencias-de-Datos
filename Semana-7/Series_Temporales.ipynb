```python
import pandas as pd

df = pd.DataFrame({
    'Mes': ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio',
            'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'],
    'Año 1': [800, 850, 900, 950, 1000, 1050, 1100, 1150, 1200, 1300, 1600, 2000],
    'Año 2': [1000, 1050, 1100, 1150, 1200, 1250, 1300, 1350, 1400, 1550, 2000, 2600]
})


media = df[['Año 1', 'Año 2']].mean()

df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
fig, ax = plt.subplots(figsize=(10, 5))

sns.lineplot(x='Mes', y='Año 1', data=df, marker='o', linewidth=2, color='teal', label='Año 1', ax=ax)
sns.lineplot(x='Mes', y='Año 2', data=df, marker='o', linewidth=2, color='rebeccapurple', label='Año 2', ax=ax)

ax.axhline(media['Año 1'], color='teal', linestyle='--', linewidth=1.5, label=f'Media Año 1 ({media["Año 1"]:.1f})')
ax.axhline(media['Año 2'], color='rebeccapurple', linestyle='--', linewidth=1.5, label=f'Media Año 2 ({media["Año 2"]:.1f})')

plt.title('Comparativa de Ventas por Mes', fontsize=14)
plt.xlabel('Mes', fontsize=12)
plt.ylabel('Monto ($)', fontsize=12)
plt.xticks(rotation=45)

ax.legend(
    loc='center left',
    bbox_to_anchor=(1.02, 0.5),
    frameon=True,
    facecolor='white',
    edgecolor='gray',
    fontsize=10
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df_inventario = pd.DataFrame({
    'Trimestre': ['Q1', 'Q2', 'Q3', 'Q4'],
    'Año 1': [2.1, 2.4, 2.6, 3.2],
    'Año 2': [2.3, 2.6, 2.8, 3.5],
    'Año 3': [2.6, 2.9, 3.1, 3.9]
})

q1_row = df_inventario[df_inventario['Trimestre'] == 'Q1'].iloc[0]
q4_row = df_inventario[df_inventario['Trimestre'] == 'Q4'].iloc[0]

años = ['Año 1', 'Año 2', 'Año 3']


variacion = q4_row[años] - q1_row[años]
crecimiento_pct = (variacion / q1_row[años]) * 100
promedios = df_inventario[años].mean()


print("Variación por cuatrimestres")
print(variacion.to_string())

print("\nCrecimiento porcentual por año")
for col in años:
    print(f"{col}: {crecimiento_pct[col]:.2f}%")

print("\nPromedios por año")
for col in años:
    print(f"Promedio {col}: {promedios[col]:.3f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

promedios = df_inventario[['Año 1', 'Año 2', 'Año 3']].mean()

sns.set_style("whitegrid")
fig, ax = plt.subplots(figsize=(10, 5))

colores = {
    'Año 1': 'plum',
    'Año 2': 'lightcoral',
    'Año 3': 'paleturquoise'
}

for col, color in colores.items():
    sns.lineplot(x='Trimestre', y=col, data=df_inventario, marker='o', linewidth=2, color=color, label=col, ax=ax)
    ax.axhline(promedios[col], color=color, linestyle='--', linewidth=1.5, label=f'Media {col} ({promedios[col]:.3f})')

    for i, val in enumerate(df_inventario[col]):
        ax.text(x=i, y=val + 0.04, s=f'{val}', ha='center', va='bottom', fontsize=9, fontweight='bold', color='black')

plt.title('Evolución de Rotación de Inventario por Trimestre', fontsize=14)
plt.xlabel('Trimestre', fontsize=12)
plt.ylabel('Rotaciones', fontsize=12)

ax.legend(
    loc='center left',
    bbox_to_anchor=(1.02, 0.5),
    frameon=True,
    facecolor='white',
    edgecolor='gray',
    fontsize=10
)

ax.text(0, -0.22, 'Katherine Susana Rojas Saborío',
        transform=ax.transAxes, fontsize=10, fontstyle='italic', ha='left', va='top')

plt.tight_layout()
plt.show()
```
