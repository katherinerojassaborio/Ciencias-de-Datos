```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Mes": ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio', 'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'],
    "Tienda_Fisica": [4200, 4150, 4300, 4180, 4250,4100, 4220, 4190, 4260, 4150, 4300, 4230],
    "En_Linea": [800, 870, 950, 1050, 1180, 1300, 1450, 1600, 1750, 1900, 2100, 2350]
}

df = pd.DataFrame(data)

df
```

```python
df['Total_por_mes'] = df['Tienda_Fisica'] + df['En_Linea']
df
```

```python
df['peso_canal_fisica'] = (df['Tienda_Fisica'] / df['Total_por_mes']) * 100
df['peso_canal_linea'] = (df['En_Linea'] / df['Total_por_mes']) * 100

df
```

```python
media_en_linea = (df['En_Linea']).mean()
media_fisica = (df['Tienda_Fisica']).mean()
mediana_en_linea = (df['En_Linea']).median()
mediana_fisica = (df['Tienda_Fisica']).median()

print(f"La media de ventas en línea es: {media_en_linea:.2f}")
print(f"La media de ventas en físico es: {media_fisica:.2f}")
print(f"La mediana de ventas en línea es: {mediana_en_linea:.2f}")
print(f"La mediana de ventas en físico es: {mediana_fisica:.2f}")
```

```python
df.describe()
```

```python
plt.figure(figsize=(11, 6), facecolor="lavender")
plt.plot(df['Mes'], df['peso_canal_fisica'], label="Tienda Física", color="lightseagreen", marker="o" )
plt.plot(df['Mes'], df['peso_canal_linea'], label = "Tienda en Línea", color= "darkblue", marker="o" )

plt.gca().set_facecolor('none')
plt.title("Comparación de ventas", fontweight='bold')
plt.xlabel("Mes")
plt.ylabel("Porcentaje de ventas(%)")
plt.xticks(rotation=45)
plt.legend(loc="center left", bbox_to_anchor=(1, 0.5))

for x, y in zip(df['Mes'], df['peso_canal_fisica']):
    plt.text(x, y + 1, f"{round(y, 1)}%", ha='center', va='bottom')

for x, y in zip(df['Mes'], df['peso_canal_linea']):
    plt.text(x, y - 2, f"{round(y, 1)}%", ha='center', va='top')

plt.grid(True, linestyle='-', alpha=0.6)
plt.tight_layout()
plt.show()
```
