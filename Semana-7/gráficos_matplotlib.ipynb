```python
ventas_producto = df.groupby("Producto")["Total"].sum()

ventas_producto.plot(
    kind = "bar",
    figsize = (10,5),
    color = "steelblue"
)

plt.title("Ventas totales por producto") # Título del gráfico
plt.xlabel("Producto") # Etiqueta del eje x
plt.ylabel("Ventas (₡)") # Etiqueta del eje y
plt.xticks(rotation=45)
plt.show()
```

```python
## Ejemplo clase: Evolución del total de ventas
# id venta
# total

ventas_producto = df.groupby("ID_Venta")["Total"].sum()

ventas_producto.plot(
    kind = "line",
    figsize = (10,5),
    color = "purple",
    marker = "o",
    markersize = 3,
)

plt.title("Ventas totales por producto") # Título del gráfico
plt.xlabel("ID de Venta") # Etiqueta del eje x
plt.ylabel("Total") # Etiqueta del eje y
plt.xticks(rotation=45)
plt.show()

```

```python
# Crear total de cada venta
# Agrupar ventas por vendedor
# Crear gráfico

df["Total_Venta"] = df["Cantidad"] * df["Precio_Unitario"]
ventas_por_vendedor = df.groupby("Vendedor")["Total_Venta"].sum()


ax = ventas_por_vendedor.plot(
    kind = "bar",
    figsize = (10,5),
    color = "thistle"

)
plt.ticklabel_format(style="plain", axis="y")
ax.bar_label(ax.containers[0], fmt="%.0f", padding=3)

plt.title("Ventas totales por vendedor")
plt.xlabel("Nombre del vendedor")
plt.ylabel("Total (₡)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# Crear total de cada venta
# Agrupar ventas por vendedor
# Crear gráfico

df["Total_Venta"] = df["Cantidad"] * df["Precio_Unitario"]
ventas_por_vendedor = df.groupby("Vendedor")["Total_Venta"].sum()

ax = ventas_por_vendedor.plot(
    kind = "bar",
    figsize = (10,5),
    color = "thistle"

)

barras = ax.patches
for barra in barras:
    valor = barra.get_height()

    ax.text(
        barra.get_x() + barra.get_width() / 2,
        valor,
        f"₡{valor:,.0f}",
        ha="center",
        va="bottom",
        fontsize=10,
        color="black"
    )

ax.yaxis.set_major_formatter(FuncFormatter(lambda x, pos: f"₡{x:,.0f}"))

plt.title("Ventas totales por vendedor") # Título del gráfico
plt.xlabel("Nombre del vendedor") # Etiqueta del eje x
plt.ylabel("Total") # Etiqueta del eje y
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

plt.figure(figsize=(14, 6))

ax = plt.gca()
ax.set_facecolor("#0F172A")
plt.gcf().patch.set_facecolor("#020617")

plt.plot(
    df["ID_Venta"],
    df["Total"],
    color="#38BDF8",
    linewidth=3,
    marker="o",
    markersize=5,
    markerfacecolor="#FFFFFF",
    markeredgecolor="#38BDF8",
    markeredgewidth=2
)

plt.title(
    "Evolución de las Ventas",
    fontsize=20,
    fontweight="bold",
    color="white",
    pad=20
)

plt.xlabel("ID de Venta", color="#CBD5E1")
plt.ylabel("Ventas (₡)", color="#CBD5E1")

# Formato ₡ en el eje Y
ax.yaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"₡{x:,.0f}")
)

plt.tick_params(colors="#CBD5E1")

plt.grid(
    True,
    linestyle="--",
    alpha=0.20,
    color="white"
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_color("#334155")
ax.spines["bottom"].set_color("#334155")

plt.tight_layout()
plt.show()
```
